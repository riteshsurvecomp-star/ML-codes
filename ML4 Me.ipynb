```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def function(x):
    return (x + 3) ** 2

def gradient(x):
    return 2 * (x + 3)
```

```python
x = 2
learning_rate = 0.1
iterations = 20

x_values = []
y_values = []
```

```python
for i in range(iterations):
    grad = gradient(x)

    x_values.append(x)
    y_values.append(function(x))

    x = x - learning_rate * grad

    print("Iteration:", i + 1, "x =", round(x, 6),
          "y =", round(function(x), 6))
```

```python
print("\nLocal Minimum:")
print("x =", round(x, 6))
print("y =", round(function(x), 6))
```

```python
x_range = np.linspace(-6, 4, 100)

plt.plot(x_range, function(x_range), label='y = (x + 3)^2')
plt.scatter(x_values, y_values, color='red', label='Gradient Descent')

plt.xlabel('x')
plt.ylabel('y')
plt.title('Gradient Descent')
plt.legend()
plt.grid()
plt.show()
```
